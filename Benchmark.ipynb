```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
sns.set_theme(
    style="darkgrid",
    palette="deep",        # default seaborn colors (blue/orange)
    context="notebook"     # scaling similar to your figure
)

plt.rcParams.update({
    "figure.figsize": (10, 6),
    "axes.titlesize": 12,
    "axes.labelsize": 11,
    "legend.fontsize": 10
})
```

```python
recall_dict = {
    'Gowalla'      : "0.1740 0.1470 0.1337".split(" "),
    'Movielens 1M' : "0.1034 0.1297 0.0822".split(" "),
    'Yelp'         : "0.0771 0.0617 0.0650".split(" "),
    'Amazon Books' : "0.0607 0.0655 0.0784".split(" "),
    'Steam'        : "0.1526 0.1531 0.1215".split(" "),
}

ndcg_dict = {
    'Gowalla'        : "0.1209 0.1015 0.0830".split(" "),
    'Movielens 1M'   : "0.0697 0.0877 0.0435".split(" "),
    'Yelp'           : "0.0416 0.0338 0.0280".split(" "),
    'Amazon Books'   : "0.0307 0.0345 0.0364".split(" "),
    'Steam'          : "0.0791 0.0776 0.0636".split(" ")
}
```

```python
recall_df = pd.DataFrame.from_dict(recall_dict, orient='index', columns=['NESCL', 'DiffRec', 'LightGCL']).astype("float64")
ndcg_df = pd.DataFrame.from_dict(ndcg_dict, orient='index', columns=['NESCL', 'DiffRec', 'LightGCL']).astype("float64")
```

```python
x = np.arange(recall_df.shape[0])

width = 0.25

ax = plt.subplot()

ax.bar(x-0.25, recall_df['NESCL'], width=width, label='NESCL')
ax.bar(x, recall_df['DiffRec'], width=width, label='DiffRec')
ax.bar(x+0.25, recall_df['LightGCL'], width=width, label='LightGCL')

ax.set_ylabel("Recall@20")

ax.set_xticks(x)
ax.set_xticklabels(recall_df.index, rotation=15)
ax.legend(loc='upper center')

plt.show()
```

```python
x = np.arange(ndcg_df.shape[0])

width = 0.25

ax = plt.subplot()

ax.bar(x-0.25, ndcg_df['NESCL'], width=width, label='NESCL')
ax.bar(x, ndcg_df['DiffRec'], width=width, label='DiffRec')
ax.bar(x+0.25, ndcg_df['LightGCL'], width=width, label='LightGCL')

ax.set_ylabel("NDCG@20")

ax.set_xticks(x)
ax.set_xticklabels(ndcg_df.index, rotation=15)
ax.legend()

plt.show()
```
