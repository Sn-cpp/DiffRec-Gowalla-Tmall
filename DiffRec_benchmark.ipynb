# Benchmarking tool

This notebook is used for fine-tuning DiffRec model on specific dataset. Before using this notebook, please ensure all necessary packages are installed. The list of packages can be found inside the `requirements.txt`.

You can adjust the hyperparameters grid below and the fixed settings inside the `Namespace` object if needed. 

For the arguments information, run:
 ```
python main.py --help
 ```
 inside the terminal.

````python
import itertools
import pandas as pd
import os
from argparse import Namespace
````

````python
# Use tqdm.notebook for better logging
os.environ["USE_TQDM_NOTEBOOK"] = "1"
````

````python
from benchmark import do_benchmark
````

````python
# Define the grid of hyperparameters for tuning
grid = {
    "emb_size" : [10, 128, 256],
    "dims" : ["[300]", "[200, 600]", "[1000]"],
    "sampling_steps": [10, 20]
}

keys = grid.keys()
values = grid.values()
````

````python
# Fine-tuning on the Gowalla dataset
if __name__ == "__main__":
    for config in itertools.product(*values):
        mapping = dict(zip(keys, config))

        args = Namespace(
            # Fixed settings
            cuda = True,
            topN = "[20]", # Fine-tuning using Recall@20
            lr = 0.001,
            batch_size = 256,
            epochs = 100,
            steps = 40,
            noise_scale = 0.1,
            noise_min = 0.0001,
            noise_max = 0.02,

            dataset = "gowalla",
            emb_size = mapping['emb_size'],
            dims = mapping['dims'],
            sampling_steps = mapping['sampling_steps']
        )

        do_benchmark(args)
````

````python
# Fine-tuning on the Tmall dataset
if __name__ == "__main__":
    for config in itertools.product(*values):
        mapping = dict(zip(keys, config))

        args = Namespace(
            cuda = True,
            topN = "[20]", # Fine-tuning using Recall@20
            lr = 0.001,
            batch_size = 256,
            epochs = 100,
            steps = 40,
            noise_scale = 0.1,
            noise_min = 0.0001,
            noise_max = 0.02,

            dataset = "tmall",
            emb_size = mapping['emb_size'],
            dims = mapping['dims'],
            sampling_steps = mapping['sampling_steps'],
        )

        do_benchmark(args)
````
